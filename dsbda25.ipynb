```python
import pandas as pd
import numpy as np

# -------------------------------------------------
# Load Dataset
# -------------------------------------------------

cols = [
    'Class',
    'AGE',
    'SEX',
    'STEROID',
    'ANTIVIRALS',
    'FATIGUE',
    'MALAISE',
    'ANOREXIA',
    'LIVER_BIG',
    'LIVER_FIRM',
    'SPLEEN_PALPABLE',
    'SPIDERS',
    'ASCITES',
    'VARICES',
    'BILIRUBIN',
    'ALK_PHOSPHATE',
    'SGOT',
    'ALBUMIN',
    'PROTIME',
    'HISTOLOGY'
]

df = pd.read_csv("hepatitis.csv", names=cols)

# Replace ? with NaN
df.replace('?', np.nan, inplace=True)

# Convert to numeric
df = df.apply(pd.to_numeric, errors='coerce')

# Remove missing values
df.dropna(inplace=True)

print(df.head())


# -------------------------------------------------
# a. Create subsets for different sex
# -------------------------------------------------

# 1 = Male, 2 = Female

male_data = df[df['SEX'] == 1]
female_data = df[df['SEX'] == 2]

print("\nMale Data:")
print(male_data.head())

print("\nFemale Data:")
print(female_data.head())


# -------------------------------------------------
# b. Merge two subsets
# -------------------------------------------------

merged_data = pd.concat([male_data, female_data])

print("\nMerged Data:")
print(merged_data.head())


# -------------------------------------------------
# c. Sort Data using AGE, SGOT, PROTIME
# -------------------------------------------------

sorted_data = df.sort_values(
    by=['AGE', 'SGOT', 'PROTIME']
)

print("\nSorted Data:")
print(sorted_data.head())


# -------------------------------------------------
# d. Transposing Data
# -------------------------------------------------

transposed_data = df.transpose()

print("\nTransposed Data:")
print(transposed_data.head())


# -------------------------------------------------
# e. Melting Data to long format
# -------------------------------------------------

melted_data = pd.melt(
    df,
    id_vars=['SEX'],
    value_vars=['AGE', 'SGOT', 'PROTIME']
)

print("\nMelted Data:")
print(melted_data.head())


# -------------------------------------------------
# f. Casting data to wide format
# -------------------------------------------------

wide_data = melted_data.pivot_table(
    index='SEX',
    columns='variable',
    values='value',
    aggfunc='mean'
)

print("\nWide Format Data:")
print(wide_data)
```
