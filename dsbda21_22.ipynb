```python
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------------------------
# Load Dataset
# -------------------------------------------------

cols = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'income'
]

df = pd.read_csv("adult.csv", names=cols)

# Replace ? with NaN
df.replace('?', pd.NA, inplace=True)

# Remove missing values
df.dropna(inplace=True)

print(df.head())


# -------------------------------------------------
# f. Histogram
# Objective:
# To visualize age distribution
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.hist(df['age'], bins=10)

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()


# -------------------------------------------------
# g. Dot Plot
# Objective:
# To visualize hours-per-week values
# -------------------------------------------------

plt.figure(figsize=(8,4))
plt.plot(df['hours-per-week'], 'o')

plt.title("Dot Plot of Hours Per Week")
plt.xlabel("Index")
plt.ylabel("Hours Per Week")

plt.show()


# -------------------------------------------------
# h. Bar Plot
# Objective:
# To compare income categories
# -------------------------------------------------

income_counts = df['income'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(income_counts.index, income_counts.values)

plt.title("Bar Plot of Income Categories")
plt.xlabel("Income")
plt.ylabel("Count")

plt.show()


# -------------------------------------------------
# i. Line Chart
# Objective:
# To visualize age trend
# -------------------------------------------------

plt.figure(figsize=(8,4))
plt.plot(df['age'])

plt.title("Line Chart of Age")
plt.xlabel("Index")
plt.ylabel("Age")

plt.show()


# -------------------------------------------------
# j. Histogram + Scatter Plot + Box Plot
# Objective:
# To visualize distribution and relationship
# -------------------------------------------------

# Histogram
plt.figure(figsize=(6,4))
plt.hist(df['hours-per-week'], bins=10)

plt.title("Histogram of Hours Per Week")
plt.xlabel("Hours")
plt.ylabel("Frequency")

plt.show()


# Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(df['age'], df['hours-per-week'])

plt.title("Scatter Plot of Age vs Hours")
plt.xlabel("Age")
plt.ylabel("Hours Per Week")

plt.show()


# Box Plot
plt.figure(figsize=(6,4))
plt.boxplot(df['hours-per-week'])

plt.title("Box Plot of Hours Per Week")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------------------------
# Load Dataset
# -------------------------------------------------

cols = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'income'
]

df = pd.read_csv("adult.csv", names=cols)

df.replace('?', pd.NA, inplace=True)
df.dropna(inplace=True)

print(df.head())


# -------------------------------------------------
# a. Histogram
# Objective:
# To visualize age distribution
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.hist(df['age'], bins=10)

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()


# -------------------------------------------------
# b. Pie Chart
# Objective:
# To visualize income category proportion
# -------------------------------------------------

income_counts = df['income'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    income_counts,
    labels=income_counts.index,
    autopct='%1.1f%%'
)

plt.title("Pie Chart of Income Categories")

plt.show()


# -------------------------------------------------
# c. Box Plot
# Objective:
# To detect spread and outliers in age
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.boxplot(df['age'])

plt.title("Box Plot of Age")

plt.show()


# -------------------------------------------------
# d. Scatter Plot
# Objective:
# To visualize relationship between age and hours
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.scatter(df['age'], df['hours-per-week'])

plt.title("Scatter Plot of Age vs Hours Per Week")
plt.xlabel("Age")
plt.ylabel("Hours Per Week")

plt.show()
```
