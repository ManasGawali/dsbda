```python
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------------------------
# Load and Merge Heart Datasets
# -------------------------------------------------

cols = [
    'age',
    'sex',
    'cp',
    'trestbps',
    'chol',
    'fbs',
    'restecg',
    'thalach',
    'exang',
    'oldpeak',
    'slope',
    'ca',
    'thal',
    'target'
]

# Load datasets
df1 = pd.read_csv("Cleavland.csv", names=cols)
df2 = pd.read_csv("hung.csv", names=cols)
df3 = pd.read_csv("Switzerland.csv", names=cols)

# Merge datasets
df = pd.concat([df1, df2, df3], ignore_index=True)

# Replace ? with NaN
df.replace('?', pd.NA, inplace=True)

# Convert to numeric
df = df.apply(pd.to_numeric, errors='coerce')

# Remove missing values
df.dropna(inplace=True)

print(df.head())


# -------------------------------------------------
# a. Histogram
# Objective:
# To visualize distribution of age
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.hist(df['age'], bins=10)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()


# -------------------------------------------------
# b. Dot Plot
# Objective:
# To visualize cholesterol values
# -------------------------------------------------

plt.figure(figsize=(8,4))
plt.plot(df['chol'], 'o')
plt.title("Dot Plot of Cholesterol")
plt.xlabel("Index")
plt.ylabel("Cholesterol")
plt.show()


# -------------------------------------------------
# c. Bar Plot
# Objective:
# To compare count of heart disease classes
# -------------------------------------------------

target_counts = df['target'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(target_counts.index.astype(str), target_counts.values)
plt.title("Bar Plot of Target Classes")
plt.xlabel("Target")
plt.ylabel("Count")
plt.show()


# -------------------------------------------------
# d. Line Chart
# Objective:
# To visualize age trend
# -------------------------------------------------

plt.figure(figsize=(8,4))
plt.plot(df['age'])
plt.title("Line Chart of Age")
plt.xlabel("Index")
plt.ylabel("Age")
plt.show()


# -------------------------------------------------
# e. Histogram + Scatter Plot + Box Plot
# Objective:
# To visualize spread, distribution and relationship
# between age and cholesterol
# -------------------------------------------------

# Histogram
plt.figure(figsize=(6,4))
plt.hist(df['chol'], bins=10)
plt.title("Histogram of Cholesterol")
plt.xlabel("Cholesterol")
plt.ylabel("Frequency")
plt.show()

# Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(df['age'], df['chol'])
plt.title("Scatter Plot of Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()

# Box Plot
plt.figure(figsize=(6,4))
plt.boxplot(df['chol'])
plt.title("Box Plot of Cholesterol")
plt.ylabel("Cholesterol")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------------------------
# Load and Merge Heart Datasets
# -------------------------------------------------

cols = [
    'age',
    'sex',
    'cp',
    'trestbps',
    'chol',
    'fbs',
    'restecg',
    'thalach',
    'exang',
    'oldpeak',
    'slope',
    'ca',
    'thal',
    'target'
]

df1 = pd.read_csv("Cleavland.csv", names=cols)
df2 = pd.read_csv("hung.csv", names=cols)
df3 = pd.read_csv("Switzerland.csv", names=cols)

df = pd.concat([df1, df2, df3], ignore_index=True)

# Cleaning
df.replace('?', pd.NA, inplace=True)
df = df.apply(pd.to_numeric, errors='coerce')
df.dropna(inplace=True)

print(df.head())


# -------------------------------------------------
# a. Histogram
# Objective:
# To visualize distribution of age
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.hist(df['age'], bins=10)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()


# -------------------------------------------------
# b. Pie Chart
# Objective:
# To visualize proportion of heart disease classes
# -------------------------------------------------

target_counts = df['target'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    target_counts,
    labels=target_counts.index,
    autopct='%1.1f%%'
)

plt.title("Pie Chart of Heart Disease Classes")
plt.show()


# -------------------------------------------------
# c. Box Plot
# Objective:
# To detect spread and outliers in cholesterol
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.boxplot(df['chol'])
plt.title("Box Plot of Cholesterol")
plt.ylabel("Cholesterol")
plt.show()


# -------------------------------------------------
# d. Scatter Plot
# Objective:
# To visualize relationship between age and cholesterol
# -------------------------------------------------

plt.figure(figsize=(6,4))
plt.scatter(df['age'], df['chol'])

plt.title("Scatter Plot of Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")

plt.show()


# -------------------------------------------------
# e. Add Boxplots to Scatterplot
# Objective:
# To compare distribution and relationship
# -------------------------------------------------

# Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(df['age'], df['chol'])

plt.title("Scatter Plot")
plt.xlabel("Age")
plt.ylabel("Cholesterol")

plt.show()


# Box Plot
plt.figure(figsize=(6,4))
plt.boxplot([df['age'], df['chol']])

plt.title("Box Plot of Age and Cholesterol")
plt.xticks([1,2], ['Age', 'Cholesterol'])

plt.show()
```
