```python
import pandas as pd
import numpy as np

# -------------------------------------------------
# Load Dataset
# -------------------------------------------------

df = pd.read_csv("forestfires.csv")

print(df.head())


# -------------------------------------------------
# a. Create data subsets by affected area
# -------------------------------------------------

# Create classes based on area

df['AffectedClass'] = pd.cut(
    df['area'],
    bins=[-1, 0, 50, np.inf],
    labels=['NotAffected', 'PartiallyAffected', 'MostlyAffected']
)

# Create subsets
not_affected = df[df['AffectedClass'] == 'NotAffected']
partial_affected = df[df['AffectedClass'] == 'PartiallyAffected']
mostly_affected = df[df['AffectedClass'] == 'MostlyAffected']

print("\nNot Affected:")
print(not_affected.head())

print("\nPartially Affected:")
print(partial_affected.head())

print("\nMostly Affected:")
print(mostly_affected.head())


# -------------------------------------------------
# b. Merge two subsets
# -------------------------------------------------

merged_data = pd.concat([partial_affected, mostly_affected])

print("\nMerged Data:")
print(merged_data.head())


# -------------------------------------------------
# c. Sort Data using Temperature, Wind and Area
# -------------------------------------------------

sorted_data = df.sort_values(
    by=['temp', 'wind', 'area']
)

print("\nSorted Data:")
print(sorted_data.head())


# -------------------------------------------------
# d. Transposing Data
# -------------------------------------------------

transposed_data = df.transpose()

print("\nTransposed Data:")
print(transposed_data.head())


# -------------------------------------------------
# e. Melting Data to long format
# -------------------------------------------------

melted_data = pd.melt(
    df,
    id_vars=['AffectedClass'],
    value_vars=['temp', 'wind', 'area']
)

print("\nMelted Data:")
print(melted_data.head())


# -------------------------------------------------
# f. Casting data to wide format
# -------------------------------------------------

wide_data = melted_data.pivot_table(
    index='AffectedClass',
    columns='variable',
    values='value',
    aggfunc='mean'
)

print("\nWide Format Data:")
print(wide_data)
```
