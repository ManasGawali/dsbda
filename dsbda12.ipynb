```python
import pandas as pd
```

```python
# Column names for Adult dataset
cols = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'income'
]

# Load dataset
df = pd.read_csv("adult.csv", names=cols)

print(df.head())
```

```python
# s. Create data subsets for different Country, Sex, Race

india_data = df[df['native-country'] == 'India']
male_data = df[df['sex'] == 'Male']
white_data = df[df['race'] == 'White']

print("\nIndia Data:")
print(india_data.head())

print("\nMale Data:")
print(male_data.head())

print("\nWhite Race Data:")
print(white_data.head())
```

```python
# t. Merge two subsets

merged_data = pd.concat([india_data, male_data])

print("\nMerged Data:")
print(merged_data.head())
```

```python
# u. Sort Data using age (or fnlwgt/income if needed)

sorted_data = df.sort_values(by='age')

print("\nSorted Data:")
print(sorted_data.head())
```

```python
# v. Transposing Data

transposed_data = df.transpose()

print("\nTransposed Data:")
print(transposed_data.head())
```

```python
# w. Melting Data to long format

melted_data = pd.melt(
    df,
    id_vars=['sex'],
    value_vars=['age', 'hours-per-week']
)

print("\nMelted Data:")
print(melted_data.head())
```

```python
# x. Casting data to wide format

wide_data = melted_data.pivot_table(
    index='sex',
    columns='variable',
    values='value',
    aggfunc='mean'
)

print("\nWide Format Data:")
print(wide_data)
```
