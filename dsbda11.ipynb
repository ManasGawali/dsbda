```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("movies_metadata.csv")

print(df.head())
```

```python
# m. Create data subsets for different languages

english_movies = df[df['original_language'] == 'en']
hindi_movies = df[df['original_language'] == 'hi']

print("\nEnglish Movies:")
print(english_movies.head())

print("\nHindi Movies:")
print(hindi_movies.head())
```

```python
# n. Merge two subsets

merged_data = pd.concat([english_movies, hindi_movies])

print("\nMerged Data:")
print(merged_data.head())
```

```python
# o. Sort Data using customer ratings

sorted_data = df.sort_values(by='vote_average')

print("\nSorted Data:")
print(sorted_data.head())
```

```python
# p. Transposing Data

transposed_data = df.transpose()

print("\nTransposed Data:")
print(transposed_data.head())
```

```python
# q. Melting Data to long format

melted_data = pd.melt(
    df,
    id_vars=['original_language'],
    value_vars=['vote_average', 'vote_count']
)

print("\nMelted Data:")
print(melted_data.head())
```

```python
# r. Casting data to wide format

wide_data = melted_data.pivot_table(
    index='original_language',
    columns='variable',
    values='value',
    aggfunc='mean'
)

print("\nWide Format Data:")
print(wide_data)
```
