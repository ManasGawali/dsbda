```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://store.steampowered.com/appreviews/730?json=1"

response = requests.get(url)

data = response.json()

reviews = data['reviews']

review_data = []

for r in reviews:

    review_data.append({
        'Customer_Name': r['author']['steamid'],
        'Review': r['review'],
        'Rating': r['voted_up'],
        'Comment_Tags': r['weighted_vote_score']
    })

df = pd.DataFrame(review_data)

print(df.head())

df.to_csv("steam_reviews.csv", index=False)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://webscraper.io/test-sites/e-commerce/static"

response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')

products = soup.find_all('div', class_='card thumbnail')

data = []

for p in products:

    name = p.find('a', class_='title').text.strip()

    price = p.find('span', itemprop='price').text.strip()

    description = p.find('p', class_='description').text.strip()

    reviews = p.find('p', class_='review-count').text.strip()

    rating = len(p.find_all('span', class_='ws-icon ws-icon-star'))

    data.append({
        'Customer_Name': name,
        'Review': description,
        'Rating': rating,
        'Comment_Tags': reviews,
        'Price': price
    })

df = pd.DataFrame(data)

print(df.head())

df.to_csv("webscraper_reviews.csv", index=False)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://pict.edu/"

response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')

headings = soup.find_all(['h1', 'h2', 'h3'])

data = []

for h in headings:

    data.append({
        'Customer_Name': 'Visitor',
        'Review': h.text.strip(),
        'Rating': 'N/A',
        'Comment_Tags': 'College'
    })

df = pd.DataFrame(data)

print(df.head())

df.to_csv("pict_reviews.csv", index=False)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://books.toscrape.com/"

response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')

books = soup.find_all('article', class_='product_pod')

data = []

for b in books:

    title = b.h3.a['title']

    rating = b.p['class'][1]

    price = b.find('p', class_='price_color').text

    data.append({
        'Customer_Name': title,
        'Review': price,
        'Rating': rating,
        'Comment_Tags': 'Book'
    })

df = pd.DataFrame(data)

print(df.head())

df.to_csv("books_reviews.csv", index=False)
```
