```python
import pandas as pd
```

```python
# Define column names
cols = [
    'SepalLengthCm',
    'SepalWidthCm',
    'PetalLengthCm',
    'PetalWidthCm',
    'Species'
]

# Load dataset with custom column names
df = pd.read_csv("Iris.csv", names=cols)
df
```

```python
# g. Create subsets
setosa = df[df['Species'] == 'Iris-setosa']
versicolor = df[df['Species'] == 'Iris-versicolor']
virginica = df[df['Species'] == 'Iris-virginica']
```

```python
# h. Merge subsets
merged_data = pd.concat([setosa, versicolor, virginica])
merged_data
```

```python
# i. Sort by Petal Length
sorted_data = df.sort_values(by='PetalLengthCm')
sorted_data
```

```python
# j. Transpose data
transposed_data = df.transpose()
transposed_data
```

```python
# k. Melt data
melted_data = pd.melt(df,
                      id_vars=["Species"],
                      value_vars=['PetalLengthCm', 'SepalLengthCm']
)
melted_data
```

```python
# l. Cast to wide format
wide_formated_data = melted_data.pivot_table(
    index = 'Species',
    columns = 'variable',
    values='value',
    aggfunc='mean'
)
wide_formated_data
```
