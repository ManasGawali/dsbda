```python
# Import libraries
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("facebook_metrics.csv", sep=';')
```

```python
# Display first 5 rows
print(df.head())
```

```python
# a. Create data subsets for type of post

photo_posts = df[df['Type'] == 'Photo']
status_posts = df[df['Type'] == 'Status']

print("\nPhoto Posts:")
print(photo_posts.head())

print("\nStatus Posts:")
print(status_posts.head())
```

```python
# b. Merge two subsets

merged_data = pd.concat([photo_posts, status_posts])

print("\nMerged Data:")
print(merged_data.head())
print(merged_data.tail())
print(merged_data['Type'].value_counts())
```

```python
# c. Sort Data on Page total likes

sorted_data = df.sort_values(by='Page total likes', ascending=True)

print("\nSorted Data:")
sorted_data
```

```python
# d. Transposing Data

transposed_data = df.transpose()

print("\nTransposed Data:")
print(transposed_data.head())
```

```python
# e. Melting Data to long format

melted_data = pd.melt(
    df,
    id_vars=['Type'],
    value_vars=[
        'Page total likes',
        'Total Interactions',
        'comment',
        'like',
        'share'
    ]
)

print("\nMelted Data:")
print(melted_data.head())
```

```python
# f. Casting data to wide format

wide_data = melted_data.pivot_table(
    index='Type',
    columns='variable',
    values='value',
    aggfunc='sum'
)

print(wide_data)
```
